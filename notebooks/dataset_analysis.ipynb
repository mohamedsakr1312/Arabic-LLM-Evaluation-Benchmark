## Import

```python
from datasets import load_dataset
import pandas as pd
```

## Dataset

```python
dataset = load_dataset("arbml/MArSum")
```

```python
dataset
```

```python
sample = dataset["test"][0]

print("TEXT:")
print(sample["text"])

print("\n" + "=" * 80 + "\n")

print("SUMMARY:")
print(sample["summary"])
```

```python
df = dataset["test"].to_pandas()

df.head()
```

```python
df.info()
```

```python
for i in range(3):
    print(f"\n{'='*100}")
    print(f"EXAMPLE {i+1}")
    print(f"{'='*100}")

    print("\nTEXT:")
    print(df.iloc[i]["text"])

    print("\nSUMMARY:")
    print(df.iloc[i]["summary"])
```

```python
benchmark_df = df.sample(
    n=200,
    random_state=42
).reset_index(drop=True)
```

```python
benchmark_df.head()
```

## Benchmark

```python
benchmark_df.to_csv(
    "data/benchmark/arabic_summarization_benchmark.csv",
    index=False,
    encoding="utf-8-sig"
)
```

```python
print("Missing text:", benchmark_df["text"].isna().sum())
print("Missing summaries:", benchmark_df["summary"].isna().sum())
```

```python
print("Empty texts:",
      (benchmark_df["text"].str.strip() == "").sum())

print("Empty summaries:",
      (benchmark_df["summary"].str.strip() == "").sum())
```
